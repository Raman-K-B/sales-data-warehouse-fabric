```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

print("Spark Session Started")
```

```python
sales_df = spark.read.format("excel") \
    .option("header", "true") \
    .load("Files/Sales.xlsx")

display(sales_df)
```

```python
customers_df = spark.read.format("excel") \
    .option("header", "true") \
    .load("Files/Customers.xlsx")

display(customers_df)
```

```python
products_df = spark.read.format("excel") \
    .option("header", "true") \
    .load("Files/Products.xlsx")

display(products_df)
```

```python
regions_df = spark.read.format("excel") \
    .option("header", "true") \
    .load("Files/Regions.xlsx")

display(regions_df)
```

```python
print("Sales:", sales_df.count())
print("Customers:", customers_df.count())
print("Products:", products_df.count())
print("Regions:", regions_df.count())
```

```python
sales_df.write.mode("overwrite").saveAsTable("sales_bronze")
customers_df.write.mode("overwrite").saveAsTable("customers_bronze")
products_df.write.mode("overwrite").saveAsTable("products_bronze")
regions_df.write.mode("overwrite").saveAsTable("regions_bronze")
```
