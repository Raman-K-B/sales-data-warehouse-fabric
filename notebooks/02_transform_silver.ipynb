```python
sales_df = spark.read.table("sales_bronze")
customers_df = spark.read.table("customers_bronze")
products_df = spark.read.table("products_bronze")
regions_df = spark.read.table("regions_bronze")
```

```python
print("Sales:", sales_df.count())
print("Customers:", customers_df.count())
print("Products:", products_df.count())
print("Regions:", regions_df.count())
```

```python
from pyspark.sql.functions import col, sum, when

sales_df.select([
    sum(when(col(c).isNull(), 1).otherwise(0)).alias(c)
    for c in sales_df.columns
]).show()
```

```python
customers_df.select([
    sum(when(col(c).isNull(), 1).otherwise(0)).alias(c)
    for c in customers_df.columns
]).show()
```

```python
sales_clean = sales_df.dropDuplicates()
customers_clean = customers_df.dropDuplicates()
products_clean = products_df.dropDuplicates()
regions_clean = regions_df.dropDuplicates()
```

```python
print("Sales Clean:", sales_clean.count())
print("Customers Clean:", customers_clean.count())
print("Products Clean:", products_clean.count())
print("Regions Clean:", regions_clean.count())
```

```python
from pyspark.sql.functions import col

sales_clean = sales_clean.withColumn(
    "Quantity",
    col("Quantity").cast("int")
)

sales_clean = sales_clean.withColumn(
    "SalesAmount",
    col("SalesAmount").cast("double")
)

sales_clean = sales_clean.withColumn(
    "SaleDate",
    col("SaleDate").cast("date")
)
```

```python
sales_clean.printSchema()
```

```python
sales_clean.write.mode("overwrite").saveAsTable("sales_silver")

customers_clean.write.mode("overwrite").saveAsTable("customers_silver")

products_clean.write.mode("overwrite").saveAsTable("products_silver")

regions_clean.write.mode("overwrite").saveAsTable("regions_silver")
```
