```python
sales_df = spark.read.table("sales_silver")
customers_df = spark.read.table("customers_silver")
products_df = spark.read.table("products_silver")
regions_df = spark.read.table("regions_silver")
```

```python
dim_customer = customers_df.select(
    "CustomerID",
    "CustomerName",
    "City",
    "State",
    "CustomerSegment"
).dropDuplicates()
```

```python
display(dim_customer)
```

```python
dim_product = products_df.select(
    "ProductID",
    "ProductName",
    "Category",
    "SubCategory"
).dropDuplicates()
```

```python
display(dim_product)
```

```python
dim_region = regions_df.select(
    "RegionID",
    "RegionName"
).dropDuplicates()
```

```python
display(dim_region)
```

```python
from pyspark.sql.functions import year, month, quarter

dim_date = sales_df.select("SaleDate").distinct()

dim_date = dim_date.withColumn("Year", year("SaleDate")) \
                   .withColumn("Month", month("SaleDate")) \
                   .withColumn("Quarter", quarter("SaleDate"))
```

```python
display(dim_date)
```

```python
fact_sales = sales_df.select(
    "SaleID",
    "CustomerID",
    "ProductID",
    "RegionID",
    "SaleDate",
    "Quantity",
    "SalesAmount"
)
```

```python
display(fact_sales)
```

```python
dim_customer.write.mode("overwrite").saveAsTable("DimCustomer")

dim_product.write.mode("overwrite").saveAsTable("DimProduct")

dim_region.write.mode("overwrite").saveAsTable("DimRegion")

dim_date.write.mode("overwrite").saveAsTable("DimDate")

fact_sales.write.mode("overwrite").saveAsTable("FactSales")
```

```python
print("FactSales:", fact_sales.count())
print("DimCustomer:", dim_customer.count())
print("DimProduct:", dim_product.count())
print("DimRegion:", dim_region.count())
print("DimDate:", dim_date.count())
```
